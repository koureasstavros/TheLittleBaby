```python
########################
# Add attributes components
########################

def process_add_attribute_components(folder_path, file_prefix, runtime_uuid):
    runtime_uuids = []
    if runtime_uuid == "all":
        # If a all is provided, add ithem all to the list
        runtime_uuids = get_directory_files(folder_path, file_prefix)
    else:
        # If a specific UUID is provided, add it to the list
        runtime_uuids.append(runtime_uuid)

    # Loop through each GUID to extract weights
    for runtime_uuid in runtime_uuids:
        print(runtime_uuid)
        # Load the model from a single file
        config_path_inp = f"{configs_path}/config_{runtime_uuid}.json"
        config = from_file(config_path_inp, "json")

        new_data = OrderedDict()
        for i, (key, value) in enumerate(config.items()):
            if i == 20:
                new_data["c_shuffle"] = False
            if i == 20:
                new_data["r_split"] = 0.9
            new_data[key] = value
        
        config_path_out = f"{configs_path}/config_{runtime_uuid}.json"
        towa_file(config_path_out, "json", new_data)
```

```python
########################
# Add sub attributes components
########################

def process_add_sub_attribute_components(folder_path, file_prefix, runtime_uuid):
    runtime_uuids = []
    if runtime_uuid == "all":
        # If a all is provided, add ithem all to the list
        runtime_uuids = get_directory_files(folder_path, file_prefix)
    else:
        # If a specific UUID is provided, add it to the list
        runtime_uuids.append(runtime_uuid)

    # Loop through each GUID to extract weights
    for runtime_uuid in runtime_uuids:
        print(runtime_uuid)
        # Load the model from a single file
        config_path_inp = f"{configs_path}/config_{runtime_uuid}.json"
        config = from_file(config_path_inp, "json")

        runtime = config["runtime"]
        new_runtime = OrderedDict()

        for i, (key, value) in enumerate(runtime.items()):
            if i == 1:
                if (config["c_device"] == "cpu"):
                    new_runtime["dataset"] = "shakespeare"
                else:
                    new_runtime["dataset"] = "sophocles"
            new_runtime[key] = value

        # Update the original data
        config["runtime"] = new_runtime

        config_path_out = f"{configs_path}/config_{runtime_uuid}.json"
        towa_file(config_path_out, "json", config)
```

```python
########################
# Calculate timing attribute components
########################

def process_add_timing_attribute_components(folder_path, file_prefix, runtime_uuid):
    runtime_uuids = []
    if runtime_uuid == "all":
        # If a all is provided, add ithem all to the list
        runtime_uuids = get_directory_files(folder_path, file_prefix)
    else:
        # If a specific UUID is provided, add it to the list
        runtime_uuids.append(runtime_uuid)

    # Loop through each GUID to extract weights
    for runtime_uuid in runtime_uuids:
        print(runtime_uuid)
        # Load the model from a single file
        report_path_inp = f"{outputs_path}/report_{runtime_uuid}.json"
        report = from_file(report_path_inp, "json")

        batches_train_time = 0
        batches_val_time = 0
        for line in report["batch_logs"]:
            if "train_loss" in line:
                time_str = line.split("execution_time = ")[-1]
                batches_train_time += float(time_str)
            elif "val_loss" in line:
                time_str = line.split("execution_time = ")[-1]
                batches_val_time += float(time_str)
                
        batches_train_cnt = sum("train_loss" in s for s in report["batch_logs"])
        batches_val_cnt = sum("val_loss" in s for s in report["batch_logs"])
        avg_train_time_per_batch = batches_train_time / batches_train_cnt if batches_train_cnt > 0 else 0
        avg_val_time_per_batch = batches_val_time / batches_val_cnt if batches_val_cnt > 0 else 0

        new_data = OrderedDict()
        for i, (key, value) in enumerate(report.items()):
            if i == 1:
                new_data["avg_train_time_per_batch"] = avg_train_time_per_batch
            if i == 2:
                new_data["avg_val_time_per_batch"] = avg_val_time_per_batch
            new_data[key] = value
        report_path_out = f"{outputs_path}/report_{runtime_uuid}.json"

        towa_file(report_path_out, "json", new_data)
```

```python
########################
# Identify wrong timing attribute components
########################

def process_identify_wrongtiming_attribute_components(folder_path, file_prefix, runtime_uuid):
    runtime_uuids = []
    if runtime_uuid == "all":
        # If a all is provided, add ithem all to the list
        runtime_uuids = get_directory_files(folder_path, file_prefix)
    else:
        # If a specific UUID is provided, add it to the list
        runtime_uuids.append(runtime_uuid)

    # Loop through each GUID to extract weights
    for runtime_uuid in runtime_uuids:
        report_path_inp = f"{outputs_path}/report_{runtime_uuid}.json"
        report = from_file(report_path_inp, "json")
        # Regex pattern to match execution_time values
        pattern = r"execution_time\s*=\s*(\d+(\.\d+)?)"

        # Filter batch logs with execution_time > 100
        matches = []
        for line in report["batch_logs"]:
            match = re.search(pattern, line)
            if match:
                value = float(match.group(1))
                if value > 100:
                    matches.append((line, value))

        # Print results
        for line, value in matches:
            print(f"File: {runtime_uuid} -> {line} -> {value:.4f}")
```

```python
########################
# Explore average timing attribute components
########################

def process_explore_average_timings_attribute_components(folder_path, file_prefix, runtime_uuid):
    runtime_uuids = []
    if runtime_uuid == "all":
        # If a all is provided, add ithem all to the list
        runtime_uuids = get_directory_files(folder_path, file_prefix)
    else:
        # If a specific UUID is provided, add it to the list
        runtime_uuids.append(runtime_uuid)

    # Loop through each GUID to extract weights
    for runtime_uuid in runtime_uuids:
        report_path_inp = f"{outputs_path}/report_{runtime_uuid}.json"
        report = from_file(report_path_inp, "json")

        batches_train_execution = []
        batches_val_execution = []
        for line in report["batch_logs"]:
            if "train_loss" in line:
                train_match = re.search(r"train_loss\s*=\s*([0-9.]+)", line)
                exec_match = re.search(r"execution_time\s*=\s*([0-9.]+)", line)
                if train_match and exec_match:
                    train_loss = float(train_match.group(1))
                    exec_time = float(exec_match.group(1))
                    batches_train_execution.append(exec_time)
            elif "val_loss" in line:
                val_match = re.search(r"val_loss\s*=\s*([0-9.]+)", line)
                exec_match = re.search(r"execution_time\s*=\s*([0-9.]+)", line)
                if val_match and exec_match:
                    val_loss = float(val_match.group(1))
                    exec_time = float(exec_match.group(1))
                    batches_val_execution.append(exec_time)
        
        print(f"Processing {runtime_uuid}")
        print("Training Batches Execution Time:", sum(batches_train_execution) / len (batches_train_execution))
        print("Validation Batches Execution Time:", sum(batches_val_execution) / len (batches_val_execution))
```
